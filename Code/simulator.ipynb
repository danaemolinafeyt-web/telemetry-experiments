```python
import random
```

```python
groups = []
times = []
canceled = []

control_cancels = 0
treatment_cancels = 0 

for i in range(1, 1001):
    group = random.choice(["control", "treatment"])
    groups.append(group)

    if group == "control":
        reframe_time = 24
        cancel_rate = 0.40
    else:
        reframe_time = 8
        cancel_rate = 0.05
    times.append(reframe_time)

    did_cancel = random.random() < cancel_rate
    if group == "control" and did_cancel:
        control_cancels += 1
    elif group == "treatment" and did_cancel:
        treatment_cancels += 1
    canceled.append(did_cancel)

print("control:", groups.count("control"))
print("treatment:", groups.count("treatment"))

control_rate = control_cancels / groups.count("control")
treatment_rate = treatment_cancels / groups.count("treatment")
print(f"Control cancel rate: {control_rate:.1%}")
print(f"Treatment cancel rate: {treatment_rate:.1%}")

```
